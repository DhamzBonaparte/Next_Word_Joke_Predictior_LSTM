```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv('/content/qajokes1.1.2.csv')
```

```python
jokes=df['Question']
```

```python
data=jokes[jokes.duplicated()==False]
```

## Tokenization

```python
from tensorflow.keras.preprocessing.text import Tokenizer
tokenizer=Tokenizer()
tokenizer.fit_on_texts(data)
```

```python
tokenizer.word_docs
```

```python
tokens=tokenizer.texts_to_sequences(data)
```

```python
tokens
```

```python
arr=[]
for sent in tokens:
  for i in range(2,len(sent)+1):
    arr.append(sent[:i])
```

```python
arr
```

```python
max_len=max([len(a) for a in arr])
```

```python
max_len
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
pad=pad_sequences(arr,maxlen=max_len,padding='pre')
```

```python
len(pad[0])
```

```python
x=pad[:,:-1]
```

```python
y=pad[:,-1]
```

```python
import tensorflow as tf
```

```python
train_ds, test_ds = tf.keras.utils.split_dataset(
    (x, y),
    left_size=0.8,   # Fraction of data to allocate to the training split
    shuffle=True,    # Shuffle data before splitting
    seed=42          # Reproducible random seed
)
```

```python
train_ds
```

```python
test_ds
```

```python
train_ds = train_ds.batch(64).prefetch(tf.data.AUTOTUNE)
```

```python
test_ds = test_ds.batch(64).prefetch(tf.data.AUTOTUNE)
```

```python
len(tokenizer.word_index)
```

```python
from tensorflow.keras.utils import to_categorical
y=to_categorical(y,num_classes=23656)
```

```python
from tensorflow.keras.layers import Embedding,Dense,LSTM,Dropout,Input
from tensorflow.keras import Sequential
```

```python
x.shape
```

```python
model=Sequential()

model.add(Input(shape=(40,)))

model.add(Embedding(23656,300))
model.add(LSTM(512,return_sequences=True))
model.add(Dropout(0.3))
model.add(LSTM(512))
model.add(Dropout(0.4))

model.add(Dense(512,activation='relu',kernel_initializer='he_normal'))
model.add(Dropout(0.3))

model.add(Dense(23656,activation='softmax'))
```

```python
model.summary()
```

```python
model.compile(loss='sparse_categorical_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
from tensorflow.keras.callbacks import EarlyStopping

rs=EarlyStopping(
  monitor='val_loss',
  patience=5,
  restore_best_weights=True
)
```

```python
model.fit(train_ds,epochs=20,callbacks=[rs],verbose=1,validation_data=test_ds)
```

```python
import numpy as np

def generate_joke_live(seed_text, next_words=30, top_k=5):
    print(f"Seed: {seed_text}", end="", flush=True)

    for _ in range(next_words):
        token_list = tokenizer.texts_to_sequences([seed_text])[0]
        token_list = token_list[-40:]
        padded_tokens = np.pad(token_list, (40 - len(token_list), 0), 'constant')
        padded_tokens = padded_tokens.reshape(1, 40)

        predicted_probs = model.predict(padded_tokens, verbose=0)[0]

        top_k_indices = np.argsort(predicted_probs)[-top_k:][::-1]
        top_k_probs = predicted_probs[top_k_indices]
        top_k_probs = top_k_probs / np.sum(top_k_probs)

        chosen_index = np.random.choice(top_k_indices, p=top_k_probs)

        output_word = ""
        for word, index in tokenizer.word_index.items():
            if index == chosen_index:
                output_word = word
                break

        if output_word:
            seed_text += " " + output_word
            print(f" {output_word}", end="", flush=True)

    print("\n")

generate_joke_live("What do you call a", next_words=20)
generate_joke_live("I told my wife", next_words=20)
generate_joke_live("Why did the chicken", next_words=20)
```
